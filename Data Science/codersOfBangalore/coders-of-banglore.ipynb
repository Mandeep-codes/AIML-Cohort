```python
with open("initialdata.txt",encoding='utf-8') as f:
    data = f.read()

data
```

```python
chunks = data.split("\n\n")
chunks[0]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split("\n")
    username = sep_chunk[0]
    followers = sep_chunk[1]
    following = sep_chunk[2]
    #posts = int(sep_chunk[3].split("post")[0].replace(",",""))
    engagement_rate = sep_chunk[4]
    #print(username,followers,following,posts,engagement_rate)
    return{"username":username,"followers":followers,"following":following,"engagement_rate":engagement_rate}

parse_chunk(chunks[0])
    
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)

print(all_chunks)
```

```python
import json
s = json.dumps(all_chunks)
print(s)
```
